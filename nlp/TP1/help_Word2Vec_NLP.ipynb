## Word2Vec
**Word2vec Popular embedding method that is very fast to train.**

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.decomposition import TruncatedSVD
import nltk
from nltk.corpus import brown
```

## Data

**In this example, we use the brown corpus.**

The Brown Corpus was a carefully compiled selection of current American English, totalling about a million words drawn from a wide variety of sources. 

```python
nltk.download('brown')

# Get 2 categories and 5 sentences from each
categories = ['reviews', 'romance']
documents = []

for cat in categories:
    sents = brown.sents(categories=cat)[:5]
    docs = [" ".join(sent) for sent in sents]
    documents.extend(docs)
```

```python
print(documents)
```

## Word2Vec
Before applying word2Vect, we should apply preprocessing steps (e.g., stemming, removing stop words, tokenization,  normalization (e.g., lemmatization/stemming), ... )


```python
#Tokenization: split the text into words, Used to extract words from documents
from nltk.tokenize import word_tokenize
#from sklearn.feature_extraction.text import TfidfVectorizer
def tokenization(data):
    for i in range(len(data)):
        data[i] = word_tokenize(data[i])
    print(len(data))
    return data

documents = tokenization(documents)
print(documents)
```

**We can now apply Word2Vec to the tokenized data.**

```python
from gensim.models import Word2Vec
import nltk
import gensim
from nltk.tokenize import sent_tokenize, word_tokenize
nltk.download('punkt_tab')
import warnings

warnings.filterwarnings(action='ignore')
```

We will be using the Skip-Gram architecture for this model.
min_count=1: Includes words that appear at least once.
vector_size=100: Generates word vectors of 100 dimensions.
window=5: Considers a context window of 5 words.
sg=1: Enables the Skip-Gram model (predicts context words given a target word).

```python
model= gensim.models.Word2Vec(documents, min_count=1, vector_size=100,
                                window=5, sg=1)
```

Evaluation of word similarity. We will compute the cosine similarity between word vectors to assess semantic similarity. Cosine similarity shows how closely related two words are in terms of meaning.

model.wv.similarity(word1, word2): Computes the cosine similarity between word1 and word2 based on the trained model.

```python
print("Cosine similarity between 'Orchestra' " +
      "and 'Symphony': ",
      model.wv.similarity('Orchestra', 'Symphony'))

print("Cosine similarity between 'Concerto' " +
      "and 'Orchestra': ",
      model.wv.similarity('Concerto', 'Orchestra'))
```
