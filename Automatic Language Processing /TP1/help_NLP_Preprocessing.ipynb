## Latent Semantic Analysis (LSA)
**LSA helps identify patterns (concepts) by examining how words co-occur across different documents. It is based on SVD, which reduces the dimension of the document-term matrix.**

```python
import nltk
from nltk.corpus import stopwords
from nltk.stem import SnowballStemmer
import ssl
import re
from nltk.stem import PorterStemmer, WordNetLemmatizer 
from sklearn.feature_extraction.text import TfidfVectorizer 
```

We use a simple toy example that you can change and extend

```python
body = [
    "The sun is rising?https://google.com",
    "The lazy gray koala",
    "The slow brown koala!",
    "The sun warmed my face"
]
```

**Lowercasing**

```python
# transform uppercase to lowercase 

def lowercase_trans(data):
    print(len(data))
    for i in range(len(data)):
        #print(i)
        data[i] = data[i].lower()
    return data
body = lowercase_trans(body)
print(body)
```

**Replce Noise**

```python
def remove_noise(data):
    print(len(data))
    for i in range(len(data)):
        #print(i)
        data[i] = re.sub(r'http\S+|www\S+', '', data[i])  # remove URLs
    return data
body = remove_noise(body)
print(body)

```

**removing ponctuation**

```python
#removing ponctuation
import re

def charact_remove(data):
    for i in range(len(data)):
        #data[i] = re.sub(r'[^a-z\s]', '', data[i])
        data[i] = re.sub(r'[^\w\s]', '', data[i])
    return data

body = charact_remove(body)
print(body)
```

**Tokenization**

```python
#Tokenization: split the text into words, Used to extract words from documents
from nltk.tokenize import word_tokenize
#from sklearn.feature_extraction.text import TfidfVectorizer
def tokenization(data):
    for i in range(len(data)):
        data[i] = word_tokenize(data[i])
    print(len(data))
    return data

body = tokenization(body)
print(body)
```

**Stop-word removal**

```python
# Download the stop words
nltk.download('stopwords')

def stop_word_removal(data):
    # Get English stopwords
    stop_words = set(stopwords.words('english'))
    for i in range(len(data)):
        # Remove stopwords
        data[i] = [word for word in data[i] if word not in stop_words]
    print(len(data))
    return data
body = stop_word_removal(body)

print("After removing stop words:", body)
```

**Stemming**

```python
from nltk.stem import PorterStemmer



def stemming(data):
    stemmer = PorterStemmer()
    for i in range(len(data)):
        data[i] = [stemmer.stem(word) for word in data[i]]
    print(len(data))
    return data

body_stem = stemming(body)
print(body_stem)
```

**Lemmatization**

```python
from nltk.stem import WordNetLemmatizer #Used to lemmatize words

def lemmatization(data):
    lemmatizer = WordNetLemmatizer()
    for i in range(len(data)):
        data[i] = [lemmatizer.lemmatize(word) for word in data[i]]

    return data
    
body_lem = lemmatization(body)
print(body_lem)
```

## LSA
The first step before applying LSA is to Preprocessing (e.g., stemming, removing stop words, tokenization,  normalization (e.g., lemmatization/stemming),  filtering, treatment of synonyms and antonyms )
• Document representation (most common: bag-of-words model)
• Word importance: Vector space model (VSM) →bag-of-words

```python
from sklearn.feature_extraction.text import CountVectorizer
# We use a simple CountVectorizer function
vectorizer = CountVectorizer(min_df=1, stop_words="english")
bag_words = vectorizer.fit_transform(body)
```

```python
print(bag_words)
```

```python
bag_words.todense()
```

**We can now apply LSA to the preprocessed data. .**

```python
from sklearn.decomposition import TruncatedSVD

svd = TruncatedSVD(n_components = 2)
lsa = svd.fit_transform(bag_words)
```

**Intepret the encoded matrices**

**Intepret the documents-topics matrix**

```python
import pandas as pd

encoded_topic = pd.DataFrame(lsa, columns = ["topic1", "topic2"])

encoded_topic["body"]=body
display(encoded_topic[["body", "topic1", "topic2"]])
```

**Intepret the terms-topics matrix**

```python
dictionary = vectorizer.get_feature_names_out()
dictionary
```

```python
encoded_matrix = pd.DataFrame(svd.components_, index = ["topic1", "topic2"], columns =dictionary).T
encoded_matrix
```

```python
import numpy as np

encoded_matrix["abs_topic1"]= np.abs(encoded_matrix["topic1"])
encoded_matrix["abs_topic2"]= np.abs(encoded_matrix["topic2"])
encoded_matrix.sort_values("abs_topic1", ascending=False)
```
